# Movie recommendation system
### content based

```python
import numpy as np 
import pandas as pd
import ast
```

```python
movies=pd.read_csv("Dataset/tmdb_5000_movies.csv")
credits=pd.read_csv("Dataset/tmdb_5000_credits.csv")
```

```python
movies.head(1)
```

```python
credits.head(1)
```

```python
#merge two df

movies=movies.merge(credits,on="title")

```

```python
movies.head()
```

```python
#drop unnecessary columns

movies=movies[["id","title","overview","genres","keywords","cast","crew"]]
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()  
```

```python
movies.iloc[0].genres
```

```python
#change the format of data
#fopr eg: genres
# [{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]
# to
# ["Action","Adventure","Fantasy","Science Fiction"]


def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):     #ast.literal_eval() converts the string list to list
        L.append(i["name"])
    return L
```

```python
movies["genres"]=movies["genres"].apply(convert)
```

```python
movies["keywords"]=movies["keywords"].apply(convert)
```

```python
#For cast we want only main leads

def convert2(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):   
        if counter!=3:
            L.append(i["name"])
            counter+=1
        else:
            break
    return L
```

```python
movies["cast"]=movies["cast"].apply(convert2)
```

```python
movies
```

```python
#Director from crew

def fetch_director(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):   
        if i["job"]=="Director":
            L.append(i["name"])
            counter+=1
    return L
```

```python
movies["crew"]=movies["crew"].apply(fetch_director)
```

```python
movies.head()
```

```python
#convert overview to list

movies["overview"]=movies["overview"].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
# We will create tags using overview genres keywords cast and crew

#first remove the spaces between the words like crime fighter to crimefighter
movies["genres"]=movies["genres"].apply(lambda x :[i.replace(" ","") for i in x])
movies["keywords"]=movies["keywords"].apply(lambda x :[i.replace(" ","") for i in x])
movies["cast"]=movies["cast"].apply(lambda x :[i.replace(" ","") for i in x])
movies["crew"]=movies["crew"].apply(lambda x :[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
#create tags

movies["tags"]=movies["overview"]+movies["genres"]+movies["keywords"]+movies["cast"]+movies["crew"]
```

```python
new_df=movies[["id","title","tags"]]
```

```python
new_df.head()
```

```python
#convert list to string

new_df["tags"]=new_df["tags"].apply(lambda x:" ".join(x))
```

```python
new_df["tags"][0]
```

```python
new_df["tags"]=new_df["tags"].apply(lambda x:x.lower())
```

```python
new_df.head()
```

# Vectorisation
### We will use BagOfWords

```python
#Stemming

import nltk
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():   #convert to list
        y.append(ps.stem(i))

    return " ".join(y)    #convert back to string
```

```python
new_df["tags"]=new_df["tags"].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(max_features=5000,stop_words="english")
```

```python
vectors=cv.fit_transform(new_df["tags"]).toarray()
```

```python
vectors[0]
```

```python
print(cv.get_feature_names_out())
```

```python
# Now we will caclulate cosine distance between the datapoints --> lesser distance similar movie
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
def recommend(movie):
    movie_index=new_df[new_df["title"]==movie].index[0] 
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
    
```

```python
recommend('Avatar')
```
